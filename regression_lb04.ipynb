```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

data = {
    "Week": [1, 2, 3, 4, 5],
    "Sales": [1.2, 1.8, 2.6, 3.2, 3.8]
}
df = pd.DataFrame(data)


X = df[["Week"]]
y = df["Sales"]


model = LinearRegression()
model.fit(X, y)

predictions = model.predict(X)


print("Intercept:", model.intercept_)
print("Coefficient:", model.coef_[0])

print(f"\nRegression Equation:")
print(f"Sales = {model.coef_[0]:.2f} × Week + {model.intercept_:.2f}")

print(f"\nR² Score: {model.score(X, y):.4f}")


plt.figure(figsize=(6, 4))
plt.scatter(X, y, color="green", label="Actual Data")
plt.plot(X, predictions, color="blue", linewidth=2, label="Regression Line")

plt.title("Linear Regression: Week vs Sales")
plt.xlabel("Week")
plt.ylabel("Sales")
plt.legend()
plt.grid(True)
plt.show()     
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.datasets import make_regression

print("Multiple Linear Regression\n")

# Generate a random dataset with 50 samples and 2 features
X, y = make_regression(
    n_samples=50,
    n_features=2,
    noise=0.1,
    random_state=42   # Remove this line if you want different data every run
)

# Create and train the model
model = LinearRegression()
model.fit(X, y)

# Print intercept and coefficients
print("Intercept:", model.intercept_)

print("\nCoefficients:")
for i, coef in enumerate(model.coef_, start=1):
    print(f"Coefficient of X{i}: {coef:.4f}")

# Display the regression equation
print("\nRegression Equation:")
print(
    f"y = {model.coef_[0]:.4f} × X1 + "
    f"{model.coef_[1]:.4f} × X2 + "
    f"{model.intercept_:.4f}"
)

# R² Score
print(f"\nR² Score: {model.score(X, y):.4f}")
```

```python
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.datasets import make_regression

# Generate a random regression dataset
X, y = make_regression(
    n_samples=50,
    n_features=1,
    noise=0.1,
    random_state=42   # Makes the output the same every time
)

# Plot the dataset
plt.figure(figsize=(6, 4))
plt.scatter(X, y, color="green", label="Data Points")

plt.title("Linear Regression using Random Dataset")
plt.xlabel("X")
plt.ylabel("y")

# Create and train the model
model = LinearRegression()
model.fit(X, y)

# Predict values
predictions = model.predict(X)

# Print model parameters
print("Intercept:", model.intercept_)
print("Coefficient:", model.coef_[0])

print("\nRegression Equation:")
print(f"y = {model.coef_[0]:.2f} × X + {model.intercept_:.2f}")

# R² Score
print(f"\nR² Score: {model.score(X, y):.4f}")

# Plot regression line
plt.plot(X, predictions, color="red", linewidth=2, label="Regression Line")

plt.legend()
plt.grid(True)
plt.show()
```
