```python

import numpy as np
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y = np.array([
    [0],
    [1],
    [1],
    [0]
])

np.random.seed(42)

W1 = np.random.randn(2, 2)
b1 = np.zeros((1, 2))

W2 = np.random.randn(2, 1)
b2 = np.zeros((1, 1))

learning_rate = 0.5


epochs = 10000


def sigmoid(x):
    return 1 / (1 + np.exp(-x))


def sigmoid_derivative(x):
    return x * (1 - x)
    
for epoch in range(epochs):




    hidden_input = np.dot(X, W1) + b1
    hidden_output = sigmoid(hidden_input)


    output_input = np.dot(hidden_output, W2) + b2
    output = sigmoid(output_input)




    error = y - output
    output_delta = error * sigmoid_derivative(output)
    hidden_error = np.dot(output_delta, W2.T)
    hidden_delta = hidden_error * sigmoid_derivative(hidden_output)
    W2 += learning_rate * np.dot(hidden_output.T, output_delta)
    b2 += learning_rate * np.sum(output_delta, axis=0, keepdims=True)
    W1 += learning_rate * np.dot(X.T, hidden_delta)
    b1 += learning_rate * np.sum(hidden_delta, axis=0, keepdims=True)

    # Display error
    if (epoch + 1) % 1000 == 0:
        mse = np.mean(error ** 2)
        print("Epoch:", epoch + 1, "MSE:", mse)

print("\nFinal Predictions:")

for i in range(len(X)):

    hidden_input = np.dot(X[i], W1) + b1
    hidden_output = sigmoid(hidden_input)

    output_input = np.dot(hidden_output, W2) + b2
    prediction = sigmoid(output_input)

    print(X[i], "->", round(float(prediction), 0))
```

```python

import numpy as np


X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])


y = np.array([0, 0, 0, 1])

weights = np.zeros(2)
bias = 0


learning_rate = 0.1


epochs = 10

def activation(x):
    if x >= 0:
        return 1
    else:
        return 0


for epoch in range(epochs):

    total_error = 0

    for i in range(len(X)):

     
        z = np.dot(X[i], weights) + bias

  
        prediction = activation(z)

        error = y[i] - prediction

        # Update weights and bias
        weights = weights + learning_rate * error * X[i]
        bias = bias + learning_rate * error

        total_error += abs(error)

    print("Epoch:", epoch + 1,
          "Error:", total_error,
          "Weights:", weights,
          "Bias:", bias)


print("\nFinal Predictions:")

for i in range(len(X)):
    z = np.dot(X[i], weights) + bias
    prediction = activation(z)

    print(X[i], "->", prediction)
```
