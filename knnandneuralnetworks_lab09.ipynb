```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

import matplotlib.pyplot as plt


# Load the Iris dataset
dataset = pd.read_csv("Iris.csv")


# Display shape of dataset
print(dataset.shape)


# Display first 5 records
print(dataset.head())


# Display last 2 records
print(dataset.tail(2))


# Display statistical description
print(dataset.describe())


# Select input features
X = dataset.iloc[:, 1:5].values


# Select target/output column
y = dataset.iloc[:, 5].values


# Split dataset into training and testing data
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)


# Create KNN classifier with k = 3
classifier = KNeighborsClassifier(n_neighbors=3)


# Train the KNN model
classifier.fit(X_train, y_train)


# Predict test data
y_pred = classifier.predict(X_test)


# Display confusion matrix
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))


# Display classification report
print("Classification Report:")
print(classification_report(y_test, y_pred))


# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred) * 100

print("Accuracy of our model is equal to",
      round(accuracy, 2), "%")
```

```python
from keras.models import Sequential

from keras.layers import Dense

from sklearn.model_selection import train_test_split

from sklearn.datasets import make_blobs


# Create data

X, y = make_blobs(n_samples=100, centers=2, n_features=2,random_state=0)

X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=0)


# define model

model = Sequential()

model.add(Dense(12, input_dim=2,activation='relu'))

model.add(Dense(8, activation='relu'))

model.add(Dense(1, activation='sigmoid'))


# compile model

model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])


# Fit the model

model.fit(X_train, y_train, epochs=100, batch_size=10, verbose=0)


# evaluate the model


scores = model.evaluate(X_train, y_train, verbose=0)

print("%s: %.2f%%" % (model.metrics_names[1], scores[1]*100))
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
from sklearn.preprocessing import StandardScaler


# Load dataset from CSV file
dataset = pd.read_csv("knearest csv.csv")

# Display shape of dataset
print(dataset.shape)

# Display first 5 records
print(dataset.head())

# Display last 2 records
print(dataset.tail(2))


# Select input features
X = dataset.iloc[:, :-1].values

# Select target column
y = dataset.iloc[:, 3].values

print(X)
print(y)


# Split dataset into training and testing data
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)


# Standardize the data
scaler = StandardScaler()

scaler.fit(X_train)

X_train = scaler.transform(X_train)
X_test = scaler.transform(X_test)


# Create KNN classifier
classifier = KNeighborsClassifier(n_neighbors=3)


# Train the model
classifier.fit(X_train, y_train)


# Predict test data
y_pred = classifier.predict(X_test)


# Display actual and predicted values
print("Actual values:", y_test)
print("Predicted values:", y_pred)


# Display confusion matrix
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))


# Display classification report
print("Classification Report:")
print(classification_report(y_test, y_pred))


# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred) * 100

print("Accuracy of our model is equal to",
      round(accuracy, 2), "%")


# Predict a new student
new_student = [[9.1, 85, 8]]

print("New Student:", new_student)

# Scale the new student using the same scaler
new_student_scaled = scaler.transform(new_student)

# Predict result
predicted = classifier.predict(new_student_scaled)

print("Predicted Result:", predicted)
```
